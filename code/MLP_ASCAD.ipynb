```python
# MLP ON ASCAD
import h5py
import numpy as np
from sklearn.preprocessing import StandardScaler

f = h5py.File('/content/ASCAD.h5', "r")
print(list(f.keys()))
print(list(f['Profiling_traces'].keys()))
print(list(f['Attack_traces'].keys()))

X_train=f['Profiling_traces']['traces'][:].astype('float32')
Y_train=f['Profiling_traces']['labels'][:]
print("Shape of X_train: ",X_train.shape)
print("Shape of Y_train: ",Y_train.shape)

X_test=f['Attack_traces']['traces'][:].astype('float32')
Y_test=f['Attack_traces']['labels'][:]
print("Shape of X_test: ",X_test.shape)
print("Shape of Y_test: ",Y_test.shape)

X_train = X_train[:, 45:80]
X_test  = X_test[:, 45:80]
#normalising
scalar=StandardScaler()
X_train=scalar.fit_transform(X_train)
X_test=scalar.transform(X_test)
print("X_train dtype:", X_train.dtype)
print("X_train mean: ", X_train.mean())
print("X_train std:  ", X_train.std())
print("X_train min:  ", X_train.min())
print("X_train max:  ", X_train.max())
print("Sample values:", X_train[0][:5])

# MLP

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import Adam

Y_train_cat= to_categorical(Y_train,num_classes=256)
Y_test_cat= to_categorical(Y_test,num_classes=256)

print("shape of Y_train_cat:", Y_train_cat.shape)

mlp_model=Sequential([Dense(200,activation='relu',input_shape=(35,)),Dropout(0.3),Dense(200,activation='relu'),Dropout(0.3),Dense(200,activation='relu'),Dropout(0.3),Dense(256,activation='softmax')])


mlp_model.compile(optimizer=Adam(learning_rate=0.001),loss='categorical_crossentropy',
    metrics=['accuracy'])

mlp_model.summary()
h_mlp=mlp_model.fit(X_train,Y_train_cat,epochs=50,batch_size=256,validation_data=(X_test, Y_test_cat),verbose=1)
loss, acc = mlp_model.evaluate(X_test, Y_test_cat, verbose=0)
print(f"\nMLP Test Accuracy: {acc*100:.2f}%")
```
